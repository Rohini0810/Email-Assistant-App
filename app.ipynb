```python
import streamlit as st
from langchain_ollama import ChatOllama


# 1. Page Configuration
st.set_page_config(
    page_title="Local AI Email Assistant",
    page_icon="✉️",
    layout="centered",
)

st.title("✉️ Local AI Email Responder")
st.write("Turn your rough notes into a professional email, running 100% locally.")


# 2. User Inputs
tone = st.selectbox(
    "Select Email Tone",
    ["Professional", "Friendly", "Apologetic", "Direct"],
)

user_input = st.text_area(
    "Enter your rough draft or notes here:",
    height=150
)


# 3. Create the Ollama connection
#You are essentially saying:

#"LangChain, I want to communicate with this model through Ollama."

llm = ChatOllama(
    model="llama3.2:3b",
    base_url="http://localhost:11434",
)


# 4. The Core Logic
if st.button("Generate Professional Email"):

    if not user_input.strip():
        st.warning("Please enter some text to get started.")

    else:

        with st.spinner("Drafting your email..."):

            # 5. Prompt Engineering
            system_prompt = f"""
            You are an expert corporate communicator.

            Your task is to rewrite the user's rough notes
            into a well-structured, grammatically correct email.

            The tone of the email must be: {tone}.

            Do not include any explanations or pleasantries
            in your output, just the email itself.
            """

            # 6. Local Model Invocation
            try:

                response = llm.invoke([
                    ("system", system_prompt),
                    ("human", user_input)
                ])

                # 7. Displaying the Output
                st.subheader("Your Polished Email:")
                st.info(response.content)

            except Exception as e:

                st.error(
                    f"An error occurred: {e}. "
                    "Make sure your Ollama Docker container is running."
                )
```

But imagine you have 100,000 chunks.

We don't want our final system to simply dump all 100,000 into the LLM.

Instead:

Retrieve the most relevant K documents.

This is Top-K retrieval.

```python
print(similarities[:5])
```

Sort by similarity

key=lambda x: x[1]

We're telling Python:

Sort using the second item in each tuple.

Our tuple is:

(index, similarity)

   0       1

So:

x[1]

means:

similarity score

```python
ranked_policies = sorted(similarities, key=lambda x: x[1], reverse=True)
```

```python
top_k = 3

top_policies = ranked_policies[:top_k]
```

```python
for index, score in top_policies:
    print(f"{score:.4f} → {policies[index]}")
```

```python
text = """
Employees can take up to 20 days of annual leave per year.
Employees must request annual leave at least 5 working days in advance.
Emergency leave can be requested on the same day when unexpected circumstances occur.
Employees should inform their manager as soon as possible when taking emergency leave.
Employees can work remotely up to 3 days per week.
Remote work arrangements must be agreed upon with the employee's manager.
"""

splitter = RecursiveCharacterTextSplitter(
    chunk_size=150,
    chunk_overlap=50
)

chunks = splitter.split_text(text)

for i, chunk in enumerate(chunks):
    print(f"\n--- Chunk {i+1} ---")
    print(chunk)
```

Chroma eliminates manually creating embeddings,sorting and retrieving it does all by it's own

Interview Answer:

"Chroma is a vector store that stores embeddings and supports similarity-based retrieval. In our LangChain setup, we provide the embedding model to Chroma, which uses it to embed documents and queries."

Interview answer

If asked:

"Do you need an embedding model when using Chroma?"

Say:

"Yes. Chroma is a vector store; it doesn't replace the embedding model. We provide an embedding model to Chroma, and Chroma uses it to generate embeddings for documents and queries and perform similarity-based retrieval."

              CHROMA
                 │
       ┌─────────┴─────────┐
       ↓                   ↓

Embedding model       Vector storage
       │                   │
       ↓                   ↓
  
  Creates vectors     Stores/searches

```python
from langchain_chroma import Chroma
document_texts = [
    f"search_document: {policy}"
    for policy in policies
]
vector_store = Chroma.from_texts(
    texts=document_texts,
    embedding=embedding_model,
    collection_name="company_policies"
)
```

```python
question = "How many days of annual leave does an employee have?"

results = vector_store.similarity_search(
    question,
    k=3
)

for i, result in enumerate(results):
    print(f"\n--- Result {i+1} ---")
    print(result.page_content)
```

🔥 Interview answer

If someone asks:

"Explain RAG."

You can now say:

"RAG, or Retrieval-Augmented Generation, combines information retrieval with LLM generation. At query time, the user's question is embedded and used to retrieve the most relevant document chunks from a vector store. Those chunks are added to the LLM prompt as context, and the LLM generates an answer grounded in that retrieved information."
